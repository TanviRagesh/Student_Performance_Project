```python
import pandas as pd

student_df = pd.concat([math_df, port_df], ignore_index=True)
```

```python
student_df["Student_ID"] = range(1, len(student_df)+1)
factor_df["Student_ID"] = range(1, len(factor_df)+1)
```

```python
factor_df = factor_df.iloc[:len(student_df)]
```

```python
merged_df = pd.merge(
    student_df,
    factor_df,
    on="Student_ID",
    how="inner"
)
```

```python
factor_df = factor_df.rename(columns={
    "Gender":"Gender_Factors",
    "Internet_Access":"Internet_Access_Factors"
})
```
